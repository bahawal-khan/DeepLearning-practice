```python
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image
import numpy as np

model = ResNet50(weights="imagenet")
```

```python
def predict_image(img_path):

    img = image.load_img(img_path, target_size=(224,224))

    img_array = image.img_to_array(img)

    img_array = np.expand_dims(img_array, axis=0)

    img_array = preprocess_input(img_array)

    pred = model.predict(img_array)

    result = decode_predictions(pred, top=3)[0]

    print(f"\nPrediction for {img_path}")

    for i, (id, label, prob) in enumerate(result, start=1):
        print(f"{i}. {label} : {prob*100:.2f}%")
```

```python
predict_image("dog2.jpg")
predict_image("cat.jpg")
predict_image("car.jpg")
predict_image('brain.jpg')
predict_image('chest.jpg')
predict_image('macular.jpg')
predict_image('dermotoscopia.jpg')
```
