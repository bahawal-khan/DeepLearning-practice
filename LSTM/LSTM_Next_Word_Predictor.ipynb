```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([faq_text])
```

```python
tokenizer.word_index
```

```python
input_sequences = []
for sentence in faq_text.split('\n'):
 tok_sen=  tokenizer.texts_to_sequences([sentence])[0]

 for i in range(1,len(tok_sen)):
  input_sequences.append(tok_sen[:i+1])
```

```python
input_sequences
```

```python
##now find the length of maximum sentence
max_len = max([len(x) for x in input_sequences])
```

```python
##now do padding
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input = pad_sequences(input_sequences,maxlen=max_len,padding =  'pre')
```

```python
padded_input
```

```python
x = padded_input[:,:-1]
y = padded_input[:,-1]
```

```python
from tensorflow.keras.utils import to_categorical
y = to_categorical(y,num_classes = (523))
```

```python
y.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,LSTM,Dense
```

```python
model = Sequential([
    Embedding(523, 100),
    LSTM(150),
    Dense(523, activation='softmax')
])

model.build((None, 28))
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics = ['accuracy'])
```

```python
model.summary()
```

```python
model.fit(x,y,epochs = 200)
```
